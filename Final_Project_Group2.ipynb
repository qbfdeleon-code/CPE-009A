```python
from datetime import datetime

# inventory stored as nested list [category_name, [items]]
# each item follows the format: [name, stock, price]
inventory_data = [
    ["Graphics Cards", [
        ["1.RTX 4070", 10, 35000.00],
        ["2.RX 7800 XT", 5, 32000.00],
        ["3.RTX 4090", 2, 110000.00]
    ]],
    ["Processors", [
        ["1.Intel i7-13700K", 15, 24500.00],
        ["2.AMD Ryzen 7 7800X3D", 12, 22000.00],
        ["3.Intel i5-12400F", 20, 8500.00]
    ]],
    ["Power Supply", [
        ["1.750W Gold Modular", 8, 5200.00],
        ["2.850W Platinum", 4, 8900.00],
        ["3.650W Bronze", 15, 3200.00]
    ]],
    ["Memory Modules", [
        ["1.16GB DDR5 6000MHz", 25, 4800.00],
        ["2.32GB DDR5 5200MHz", 10, 8500.00],
        ["3.8GB DDR4 3200MHz", 40, 1500.00]
    ]],
    ["Input Devices", [
        ["1.Mechanical Keyboard", 12, 3500.00],
        ["2.Gaming Mouse", 18, 2200.00],
        ["3.Wireless Trackpad", 7, 4500.00]
    ]]
]

# sales_log stores all completed transactions as a list of dictionaries
# receipt_counter starts at 1000 so receipt numbers don't start at 1
sales_log = []
receipt_counter = 0


# --- Functions ---

# prints just the category names with numbers so the user can pick one
def display_category_menu():
    print("\nPlease select a category:")
    for i, item in enumerate(inventory_data, start=1):
        print(f"  {i}. {item[0]}")

# shows the full details (model, stock, price) of items inside a category
def show_inventory_details(cat_idx):
    category = inventory_data[cat_idx]
    print(f"\n- - - {category[0].upper()} - - -")
    print(f"{'Model':<25} {'Stock':<10} {'Price':<10}")
    print("-" * 50)
    for model in category[1]:
        name, stock, price = model
        print(f"{name:<25} {stock:<10} ₱{price:,.2f}")

# reusable loop for going back to the main menu
# placed in a function to avoid repeating the same block in every menu option
def return_to_menu():
    while True:
        back = input("\nEnter 0 to return to the Main Menu: ")
        if back == '0':
            break
        else:
            print("Invalid input.")


# --- Main Program ---

# keeps running until the user chooses to exit
while True:
    print("\n" + "=" * 45)
    print("        INVENTORY MANAGEMENT SYSTEM")
    print("=" * 45)
    print("  1. View Inventory")
    print("  2. Process Sale")
    print("  3. Restock Inventory")
    print("  4. Sales History")
    print("  5. Exit")

    choice = input("\nPlease enter your selection (1-5): ")

    # using match-case here instead of if-elif chain, cleaner for menu handling
    match choice:

        case '1':
            print("\n" + "-" * 30)
            print("         Inventory")
            print("-" * 30)

            display_category_menu()

            sub_choice = input("\nPlease select a category (1-5): ")

            if sub_choice.isdigit() and 1 <= int(sub_choice) <= len(inventory_data):
                show_inventory_details(int(sub_choice) - 1)
            else:
                print("Invalid selection.")

            return_to_menu()

        case '2':
            print("\n" + "-" * 31)
            print("         Process Sale")
            print("-" * 31)


            cart = []  # temporary list to hold items before checkout

            while True:
                display_category_menu()
                print(f"  {len(inventory_data) + 1}. Proceed to Checkout")

                cat_input = input("\nPlease select a category or choose Checkout: ")

                # if user picks the checkout option
                if cat_input == str(len(inventory_data) + 1):
                    if not cart:
                        print("Your cart is currently empty.")
                    else:
                        # generate receipt details
                        receipt_counter += 1
                        receipt_no = f"RCP-{receipt_counter}"
                        now = datetime.now().strftime("%B %d, %Y  %I:%M %p")
                        grand_total = sum(item["subtotal"] for item in cart)

                        # save the transaction to sales log
                        sales_log.append({
                            "receipt_no": receipt_no,
                            "date": now,
                            "items": cart[:],
                            "grand_total": grand_total
                        })

                        # print the receipt
                        print("\n" + "*" * 48)
                        print("           Official Sales Receipt")
                        print("*" * 48)
                        print(f"  Receipt No. : {receipt_no}")
                        print(f"  Date & Time : {now}")
                        print("*" * 48)
                        print(f"{'Item':<25} {'Qty':<6} {'Price':<11} {'Subtotal':<11}")
                        print("*" * 48)
                        for item in cart:
                            print(f"{item['model']:<26} {item['qty']:<6} ₱{item['price']:,.2f}{'':>3} ₱{item['subtotal']:,.2f}")
                        print("*" * 48)
                        print(f"{'Total Amount Due':<30} ₱{grand_total:,.2f}")
                        print("*" * 48)
                        print("       Thank you for your purchase!")
                        print("*" * 48)
                    break

                # validate category input
                if not cat_input.isdigit() or not (1 <= int(cat_input) <= len(inventory_data)):
                    print("Invalid selection. Please enter a valid category number.")
                    continue

                cat_idx = int(cat_input) - 1
                category = inventory_data[cat_idx]
                items = category[1]

                print(f"\n- - - {category[0].upper()} - - -")
                print(f"{'Model':<25} {'Stock':<10} {'Price':<10}")
                print("-" * 50)
                for i, model in enumerate(items, start=1):
                    name, stock, price = model
                    print(f"{i}. {name:<23} {stock:<10} ₱{price:,.2f}")

                prod_input = input("\nPlease enter the product number: ")
                if not prod_input.isdigit() or not (1 <= int(prod_input) <= len(items)):
                    print("Invalid product selection. Please try again.")
                    continue

                prod_idx = int(prod_input) - 1
                name, stock, price = items[prod_idx]

                qty_input = input(f"Please enter the desired quantity (Available stock: {stock}): ")
                if not qty_input.isdigit() or int(qty_input) <= 0:
                    print("Invalid quantity. Please enter a positive whole number.")
                    continue

                qty = int(qty_input)
                if qty > stock:
                    print(f"Insufficient stock. Only {stock} unit(s) are currently available.")
                    continue

                # calculate subtotal and add item to cart
                subtotal = qty * price
                cart.append({
                    "model": name,
                    "qty": qty,
                    "price": price,
                    "subtotal": subtotal
                })

                # deduct purchased quantity from inventory
                inventory_data[cat_idx][1][prod_idx][1] -= qty
                print(f"\nItem successfully added to cart: {name} x{qty} = ₱{subtotal:,.2f}")

            return_to_menu()

        case '3':
            print("\n" + "-" * 30)
            print("      Restock Inventory")
            print("-" * 30)

            display_category_menu()

            cat_input = input("\nPlease select a category to restock (1-5): ")

            if not cat_input.isdigit() or not (1 <= int(cat_input) <= len(inventory_data)):
                print("Invalid selection. Please enter a number between 1 and 5.")
            else:
                cat_idx = int(cat_input) - 1
                category = inventory_data[cat_idx]
                items = category[1]

                print(f"\n- - - {category[0].upper()} - - -")
                print(f"{'Model':<25} {'Current Stock':<10}")
                print("-" * 40)
                for i, model in enumerate(items, start=1):
                    name, stock, price = model
                    print(f"{i}. {name:<23} {stock}")

                prod_input = input("\nPlease select a product to restock: ")
                if not prod_input.isdigit() or not (1 <= int(prod_input) <= len(items)):
                    print("Invalid product selection. Please try again.")
                else:
                    prod_idx = int(prod_input) - 1
                    name, stock, price = items[prod_idx]

                    qty_input = input(f"Please enter the quantity to restock (Maximum: 10 units): ")

                    if not qty_input.isdigit():
                        print("Invalid input. Please enter a numeric value.")
                    else:
                        qty = int(qty_input)

                        if qty <= 0:
                            print("Invalid quantity. The value must be greater than zero.")
                        elif qty > 10:
                            print("Restock limit exceeded. A maximum of 10 units may be added per transaction.")
                        else:
                            # update the stock in inventory_data
                            inventory_data[cat_idx][1][prod_idx][1] += qty
                            new_stock = inventory_data[cat_idx][1][prod_idx][1]
                            print(f"\nRestock Successful.")
                            print(f"  Product  : {name}")
                            print(f"  Added    : +{qty} unit(s)")
                            print(f"  New Stock: {new_stock} unit(s)")

            return_to_menu()

        case '4':
            print("\n" + "-" * 30)
            print("       Sales History")
            print("-" * 30)

            if not sales_log:
                print("No sales transactions have been recorded yet.")
            else:
                # loop through each saved transaction and display it
                for record in sales_log:
                    print("\n" + "*" * 50)
                    print(f"Receipt : {record['receipt_no']}  |  {record['date']}")
                    print(f"{'Item':<15} {'Qty':<3} {'Subtotal':<5}")
                    print("*" * 50)
                    for item in record["items"]:
                        print(f"{item['model']:<25} {item['qty']:<6} ₱{item['subtotal']:,.2f}")
                    print("*" * 50)
                    print(f"{'TOTAL':<32} ₱{record['grand_total']:,.2f}")
                    print("*" * 50)

            return_to_menu()

        case '5':
            print("\nExiting the Program. Goodbye!")
            break

        # handles any input that isn't 1-5
        case _:
            print("Invalid selection. Please enter a number from 1 to 5.")
```
