Example 1

```python
week = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]

for x in week:
  print(x)
```

Example 2 (The break statement)

```python
week = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
for x in week:
  print(x)
  if x == "Thursday":
    break
```

```python
week = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
for x in week:
  print(x)
  if x >= "Thursday":
    break
```

Example 4(Looping through a string)

```python
for x in "week":
  print(x)
```

The range() Function

```python
#Example:
for x in range(6):
  print(x)

for x in range(2, 6):
  print(x)
```

```python
for x in range (2, 7, 2):
  print(x)
```

Nested Loop



```python
adjective = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adjective:
  for y in fruits:
    print(x, y)
```

While Loop/Statement

```python
#Example 5

i = 1
while i < 6:
  print(i)
  i+=1 #i+1 Assignement operator
```

Break statement in while loop

```python
i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i += 1
```

Continue Statement

```python
i = 0
while i < 6:
  i += 1
  if i == 3:
    continue
  print(i)
```

```python
i = 0
while i < 6:
  i += 1
  if i == 3:
    print(i)
    continue
```

The else Statement

```python
i = 1
while i<6:
  print(i)
  i += 1
else:
  print("i is no longer less than 6")
```
