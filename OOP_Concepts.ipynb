```python
class Car:
  def __init__(self, name, mileage):
    self.name = name
    self.mileage = mileage
  def description(self):
    return f"The {self.name} car gives the milage of {self.mileage} km/1"
car1 = Car("Toyota", 2000)
car2 = Car("Honda", 3000)
print(car1.description())
print(car2.description())
#
```

```python
class Animals:
  def __init__(self, name, habitat, temperature):
    self.name = name
    self.habitat = habitat
    self.temperature = temperature
  def feature(self):
    return f"The {self.name} which has a body temperature {self.temperature} lives in {self.habitat}"

mamals = Animals("Dog", "Land", "warm")
birds = Animals("Sparrow", "Air", "warm")
reptiles = Animals("snake", "terrestrial", "cold")
print(mamals.name)
print(mamals.habitat)
print(mamals.temperature)
print(mamals.feature())
print(birds.feature())
print(reptiles.feature())
```

```python
class Shapes:
  def __init__(self, name, sides):
    self.name = name
    self.sides = sides

  def description(self):
      return f"The {self.name} has {self.sides} sides"

  def computeArea(self):
    base = 3
    height = 4
    if self.name == "Triangle":
      AreaT = (0.5 * base * height)
      return AreaT

    elif self.name == "Circle":
      r = 1
      AreaC = 3.14159 * (r**2)
      return AreaC

    elif self.name == "Square":
      side = 4
      AreaS = side**2
      return AreaS

    elif self.name == "Rectangle":
      length = 3
      width = 4
      AreaR = (length * width)
      return AreaR

shape1 = Shapes("Triangle", 3)
shape2 = Shapes("Cirlce", 0)
shape3 = Shapes("Square", 4)
shape4 = Shapes("Rectangle", 4)


print(shape1.description())
print(shape2.description())
print(shape3.description())
print(shape4.description())

print(f"Area of {shape1.name}: {shape1.computeArea()}")
print(f"Area of {shape2.name}: {shape2.computeArea()}")
print(f"Area of {shape3.name}: {shape3.computeArea()}")
print(f"Area of {shape4.name}: {shape4.computeArea()}")
```
