Calculating Final Grade

```python
print(" ----- Final Grade in a CpE Program -----\n")

# Prelim Grade
x = float(input("Enter Prelim Hands-on Activity grade: "))
y = float(input("Enter Prelim Quiz grade: "))
z = float(input("Enter Prelim Assignment grade: "))

# Calculate Prelim CS
x = x * .5
y = y * .3
z = z * .2
p_cs = x + y + z
print("Prelim CS is: ", p_cs)

a = float(input("Enter Prelim Exam grade: "))
b = p_cs # Using the calculated CS
a = a * .5
b = b * .5
prelim_grade = a + b
print("Prelim Grade is: ", prelim_grade)

# Midterm Grade
x = float(input("\nEnter Midterm Hands-on Activity grade: "))
y = float(input("Enter Midterm Quiz grade: "))
z = float(input("Enter Midterm Assignment grade: "))

# Calculate Midterm CS
x = x * .5
y = y * .3
z = z * .2
m_cs = x + y + z
print("Midterm CS is: ", m_cs)

a = float(input("Enter Midterm Exam grade: "))
b = m_cs # Using the calculated CS

a = a * .5
b = b * .5
# Midterm Grade = 1/3 Prelim + 2/3
midterm_grade = (1/3 * prelim_grade) + (2/3 * (a + b))
print("Midterm Grade is: ", midterm_grade)

# Final Grade
x = float(input("\nEnter Final Hands-on Activity grade: "))
y = float(input("Enter Final Quiz grade: "))
z = float(input("Enter Final Assignment grade: "))

# Calculate Final CS
x = x * .5
y = y * .3
z = z * .2
f_cs = x + y + z
print("Final CS is: ", f_cs)

a = float(input("Enter Final Exam grade: "))
b = f_cs # Using the calculated CS

a = a * .5
b = b * .5
# Final Grade = 1/3 Midterm + 2/3
final_grade = (1/3 * midterm_grade) + (2/3 * (a + b))
print("Final Grade is: ", final_grade)

```
