```python
# Calculate total marks and percentage

grade = 0

print("Please enter your grade:" )
grade = int(input())

if grade >= 90:
  print('A+, Outstanding!')

elif grade >= 80 <=89:
  print('A, Excellent!')

elif grade >= 70 <=79:
  print('B+, Very Good.')

elif grade >= 60 <=69:
  print('B, Good.')

elif grade >= 50 <=59:
  print('C, Not bad.')

elif grade >= 40 <=49:
  print('D, Satisfactory.')

elif grade >=0 <=39:
  print('F(Fail), Needs Improvement.')

else:
  print('Invalid Grade')
```
